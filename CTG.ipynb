```python
# ========================
# STEP 1: Import libraries
# ========================
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.ensemble import RandomForestClassifier
import joblib
from google.colab import files

# ========================
# STEP 2: Load dataset
# ========================
data = pd.read_csv('/fetal_health.csv')  # Or upload via Colab if needed

# Select features
selected_features = [
    "baseline value",
    "accelerations",
    "fetal_movement",
    "uterine_contractions",
    "abnormal_short_term_variability",
    "mean_value_of_short_term_variability",
    "prolongued_decelerations",
    "histogram_mode",
    "histogram_median",
    "histogram_variance"
]

X = data[selected_features]
y = data["fetal_health"]

# ========================
# STEP 3: Scale features
# ========================
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# Train/test split
X_train, X_test, y_train, y_test = train_test_split(
    X_scaled, y, test_size=0.3, random_state=42
)

# ========================
# STEP 4: Train model
# ========================
model = RandomForestClassifier(random_state=42)
model.fit(X_train, y_train)

print(f"Model Accuracy: {model.score(X_test, y_test):.4f}")

# ========================
# STEP 5: Save model & scaler
# ========================
joblib.dump(model, "fetal_health_model.pkl")
joblib.dump(scaler, "scaler.pkl")

# ========================
# STEP 6: Download to system
# ========================
files.download("fetal_health_model.pkl")
files.download("scaler.pkl")
```

```python
import tkinter as tk
from tkinter import ttk, messagebox
import numpy as np
import joblib

# ========================
# Load model and scaler
# ========================
model = joblib.load("/content/fetal_health_model.pkl")
scaler = joblib.load("/content/scaler.pkl")

# Feature names (same order used in training)
selected_features = [
    "baseline value",
    "accelerations",
    "fetal_movement",
    "uterine_contractions",
    "abnormal_short_term_variability",
    "mean_value_of_short_term_variability",
    "prolongued_decelerations",
    "histogram_mode",
    "histogram_median",
    "histogram_variance"
]

# ========================
# Predict Function
# ========================
def predict():
    try:
        # Get values from entry boxes
        values = [float(entry_vars[feat].get()) for feat in selected_features]

        # Scale input
        values_scaled = scaler.transform([values])

        # Predict
        prediction = model.predict(values_scaled)[0]

        # Map prediction to meaning
        labels = {
            1: "Normal",
            2: "Suspect",
            3: "Pathological"
        }

        messagebox.showinfo("Prediction Result", f"Fetal Health: {labels[prediction]}")

    except ValueError:
        messagebox.showerror("Input Error", "Please enter valid numeric values.")

# ========================
# Build GUI
# ========================
root = tk.Tk()
root.title("Fetal Health Prediction")
root.geometry("500x600")

# Heading
title_label = tk.Label(root, text="Fetal Health Prediction System", font=("Arial", 16, "bold"))
title_label.pack(pady=10)

# Frame for form
form_frame = tk.Frame(root)
form_frame.pack(pady=10)

entry_vars = {}

for i, feat in enumerate(selected_features):
    label = tk.Label(form_frame, text=feat, font=("Arial", 10))
    label.grid(row=i, column=0, padx=10, pady=5, sticky="w")

    var = tk.StringVar()
    entry = ttk.Entry(form_frame, textvariable=var, width=25)
    entry.grid(row=i, column=1, padx=10, pady=5)

    entry_vars[feat] = var

# Predict button
predict_button = tk.Button(root, text="Predict", command=predict, bg="green", fg="white", font=("Arial", 12, "bold"))
predict_button.pack(pady=20)

# Exit button
exit_button = tk.Button(root, text="Exit", command=root.quit, bg="red", fg="white", font=("Arial", 12, "bold"))
exit_button.pack(pady=10)

root.mainloop()
```

```python
import ipywidgets as widgets
from IPython.display import display
import numpy as np
import joblib

# Load model and scaler
try:
    model = joblib.load("/content/fetal_health_model.pkl")
    scaler = joblib.load("/content/scaler.pkl")
except FileNotFoundError:
    print("Model or scaler file not found. Please run the previous cell to train and save them.")

# Feature names (same order used in training)
selected_features = [
    "baseline value",
    "accelerations",
    "fetal_movement",
    "uterine_contractions",
    "abnormal_short_term_variability",
    "mean_value_of_short_term_variability",
    "prolongued_decelerations",
    "histogram_mode",
    "histogram_median",
    "histogram_variance"
]

# Create input widgets
input_widgets = {}
for feat in selected_features:
    input_widgets[feat] = widgets.FloatText(description=feat)

# Output widget
output_widget = widgets.Output()

# Predict function
def predict(b):
    with output_widget:
        output_widget.clear_output()
        try:
            # Get values from input widgets
            values = [input_widgets[feat].value for feat in selected_features]

            # Scale input
            values_scaled = scaler.transform([values])

            # Predict
            prediction = model.predict(values_scaled)[0]

            # Map prediction to meaning
            labels = {
                1: "Normal",
                2: "Suspect",
                3: "Pathological"
            }

            print(f"Fetal Health Prediction: {labels[prediction]}")

        except ValueError:
            print("Please enter valid numeric values.")
        except NameError:
             print("Model or scaler not loaded. Please run the cell above.")


# Create predict button
predict_button = widgets.Button(description="Predict")
predict_button.on_click(predict)

# Arrange widgets
input_box = widgets.VBox([input_widgets[feat] for feat in selected_features])
controls = widgets.VBox([predict_button, output_widget])
dashboard = widgets.VBox([input_box, controls])

# Display the dashboard
display(dashboard)
```
